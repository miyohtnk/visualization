## Making a basic figure with matplotlib

```python
#important libraries to load
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613) #make it reproducible
x = np.arange(50) #create 1D array with 50 elements
y= np.random.randint(0,100,50) #I want random int from 0 to 100 and we want 50 y values to be generated
```

```python
fig, ax = plt.subplots(figsize= (5,3)) #create plot with x=5 by y=3 size
ax.scatter(x,y) #scatterplot of our X and Y data
```

```python
#Start modifying the figure, now in bar plot instead of scatterplot
fig, ax = plt.subplots(figsize = (5,3,))
ax.bar(x,y)
```

```python
fig, ax = plt.subplots(figsize = (5,3,))
ax.barh(x,y) #horizontal version of bar
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y) #line plot
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.hist(y) #x and y is error. We only need y values
```

```python
#Make a complete plot with axes titles and plot title
fig, ax = plt.subplots( figsize = (5,3))
ax.plot(x,y)
ax.set_title("Total growth over time")
ax.set_ylabel("Total growth")
ax.set_xlabel('Years since start')
fig.tight_layout() #keep it within bounds, especially with long titles
```

```python
#make dictionarries for modifiyng labels
font1 = {'family': 'sans-serif', 'color':'blue',
         'size': 20}
font2 = {'family':'monospace', 'color':'green',
         "size": 14}
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y)

ax.set_title("Total growth over time", fontdict=font1, loc='left')
ax.set_ylabel('Total growth', fontdict=font2)
ax.set_xlabel("Years since start", fontdict=font2)
```

## Modifying our data points

```python
fig, ax = plt.subplots(figsize = (5,3))

ax.scatter(x,y,
           marker= "*",
           color= 'indigo')
```

```python
fig, ax = plt.subplots( figsize = (5,3))

ax.plot(x,y,
        marker = '*',
        color= 'indigo',
        linestyle= '--',
        linewidth = 0.5)
        
```

```python
fig, ax = plt.subplots( figsize = (5,3))

ax.plot(x,y,
        marker= '*',
        color = "#7425b9",
        linestyle = '--',
        linewidth = 0.5)
```

```python
fig, ax = plt.subplots( figsize = (5,3))

ax.plot(x,y,
        marker= '*',
        markersize = 12,
        linestyle = '--',
        color= "#7425b9",
        linewidth = 0.5,
        markeredgecolor= "#fa9359",
        markerfacecolor = '#000000')
ax.grid(axis = 'y', color= 'blue', linewidth= 0.2,
        linestyle= '--')
```
