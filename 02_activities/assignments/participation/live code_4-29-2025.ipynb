```python
#load libraries
import numpy as np
import matplotlib.pyplot as plt
import PIL
import scipy
import requests
import pandas as pd
```

```python
#sample data
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)
```

```python
#line plot with botth y variables on same axes
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y1)
ax.plot(x,y2)
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y1, label= 'person 1')
ax.plot(x,y2, label= "person 2")
ax.legend(loc='lower right')
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y1, label= 'person 1')
ax.plot(x,y2, label = 'person 2')
ax.legend(loc= 'lower right',
          frameon= True,
          fontsize = 12,
          ncol = 2,
          shadow= True)
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y1, label = "person 1")
ax.plot(x,y2, label = "person 2")

ax.legend(loc = "upper left",
          bbox_to_anchor= (1,1)) #upper right would 
                                #mean the legend would be inside the plot
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = 'person 1')
ax.scatter(x,y2, label = "person 2")
ax.legend(loc = 'lower right')
ax.text(10,95, "This value is important!") # location x=10, y=95, beginning of text
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = 'person 1')
ax.scatter(x,y2, label = "person 2")
ax.legend(loc = 'lower right')
ax.text(10,95, "This value is important!",
        ha = 'center',
        color = 'red',
        size = 20)#ha= horizontal alignment, so x=10, y=95 is in middle
```

```python
fig, ax = plt.subplots()
ax.axis([0,10,0,10])

ax.text(1,5, ".  Data:(1,5)", transform = ax.transData)
ax.text(0.5, 0.1, ".  Axes:(0.5, 0.1)", transform = ax.transAxes)
ax.text(0.2, 0.2, ".  Figure:(0.2, 0.2)", transform=fig.transFigure)
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = 'person 1')
ax.scatter(x,y2, label = "person 2")
ax.legend(loc = 'lower right')

ax.annotate("This is important!", xy= (10,95), xytext= (20,94),
            arrowprops= dict( arrowstyle = "wedge", color= 'hotpink'))
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label= "person 1")
ax.scatter(x,y2, label = "person 2")
ax.legend(loc = "lower right")

ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_formatter(plt.NullFormatter()) #leaves the little lines
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label= "person 1")
ax.scatter(x,y2, label = "person 2")
ax.legend(loc = "lower right")

ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_locator(plt.NullLocator()) #removes all
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = "person 1")
ax.scatter(x,y2, label = 'person 2')
ax.legend(loc = 'lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(3))
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = "person 1")
ax.scatter(x,y2, label = 'person 2')
ax.legend(loc = 'lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(5))
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = "person 1")
ax.scatter(x,y2, label = 'person 2')
ax.legend(loc = 'lower right')
ax.xaxis.set_major_locator(plt.MultipleLocator(5)) #every 5
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = "person 1")
ax.scatter (x,y2, label = 'person 2')
ax.legend(loc= "lower right")

plt.xticks(rotation = 45, ha= 'center') #if you try right, then the will move from the tick
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = "person 1")
ax.scatter (x,y2, label = 'person 2')
ax.legend(loc= "lower right")

plt.xlabel(xlabel= 'Shiny New X Axis!', font= 'serif', color='indigo')
plt.xticks(rotation = 45, ha= 'center') #if you try right, then the will move from the tick
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = "person 1")
ax.scatter (x,y2, label = 'person 2')
ax.legend(loc= "lower right")

font1 = {"family": "serif", 'color': "indigo"}
plt.xlabel(xlabel= 'Shiny New X Axis!', fontdict = font1, fontsize = 18)
```
